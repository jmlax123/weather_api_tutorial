```python
def get_city_data(city, api_key):
    
    current_url = f"{base_url}/current.json?key={api_key}&q={city}"
    forecast_url = f"{base_url}/forecast.json?key={api_key}&q={city}&days=3"
    astro_url = f"{base_url}/astronomy.json?key={api_key}&q={city}&dt={today}"
    sports_url = f"{base_url}/sports.json?key={api_key}&q={city}"

    current_response = requests.get(current_url)
    forecast_response = requests.get(forecast_url)
    astro_response = requests.get(astro_url)
    sports_response_raw = requests.get(sports_url) 

    if current_response.status_code == 200 and forecast_response.status_code == 200:
        data = current_response.json()
        forecast_data = forecast_response.json()
        astro_data = astro_response.json()
        sports_response = sports_response_raw.json()

        location = data['location']
        current = data['current']
        astro = astro_data['astronomy']['astro']        

        dt_obj = datetime.strptime(location['localtime'], "%Y-%m-%d %H:%M")
        readable_time = dt_obj.strftime("%B %d, %Y, %I:%M %p")
     
        forecast = forecast_data['forecast']['forecastday']
        forecast_summary = []
        for day in forecast:
            date_obj = datetime.strptime(day['date'], '%Y-%m-%d')
            formatted_date = date_obj.strftime('%A, %B %d')
            forecast_summary.append(f"{formatted_date}: High {day['day']['maxtemp_f']}F / Low {day['day']['mintemp_f']}F - {day['day']['condition']['text']}")

        current = data['current']
        location = data['location']
        temp_f = current['temp_f']
        temp_c = current['temp_c']
        wind_mph = current['wind_mph']
        humidity = current['humidity']
        sunrise = astro['sunrise']
        sunset = astro['sunset']
        
        print(f"Weather for {location['name']}, {location['region']}")
        print(f"The current timezone is: {location['tz_id']}")
        print(f"The local date and time is: {readable_time}")
        print(f"Temperature: {temp_f} °F / {temp_c} °C")
        print(f"Humidity: {humidity}%")
        print(f"Wind Speed: {wind_mph} mph")

        print(f"\nThe sun rises at {sunrise} and sets at {sunset}.")
        print(f"The phase of the moon is {astro['moon_phase']} ({astro['moon_illumination']}% illuminated).")

        print(f"\n--- 3-Day Forecast for {location['name']} ---")
        print("\n".join(forecast_summary))

        sports_found = False
        for sport in ['football', 'cricket', 'golf']:
            if sport in sports_response and sports_response[sport]:
                sports_found = True
                print(f"\n--- Upcoming {sport.capitalize()} ---")
                for event in sports_response[sport]:
                    print(f"Match: {event['match']} at {event['stadium']} ({event['start']})")

        if not sports_found:
            print("\nNo major upcoming sports events found for this city.")

        weather_df = pd.DataFrame([{
            'City': location['name'],
            'Timezone': location['tz_id'],
            'Local_Time': readable_time,
            'Temperature': temp_f,
            'Humidity Percentage': humidity,
            'Wind Speed': wind_mph,
            'Sunrise': sunrise,
            'Sunset': sunset
        }])
        return weather_df

    else:
        print(f"Error fetching data. Status: {current_response.status_code}")

get_city_data("Louisville", "PASTE_YOUR_KEY_HERE")
```

```python
df = get_city_data("Louisville", "PASTE_YOUR_KEY_HERE")
df.to_csv('weather_data.csv', index=False)

print("File 'weather_data.csv' has been created!")
```
